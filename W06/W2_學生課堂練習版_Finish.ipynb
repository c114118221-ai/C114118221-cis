## Mission 5：條件判斷與自訂函式

**Knowledge Point**：`def`、參數、`return`、`if / elif / else`、縮排

函式把一段可重複使用的規則取名。`score` 是輸入參數；`return` 傳回結果。Python 用縮排表示哪些指令屬於函式或條件區塊。

**執行前預測**：輸入 82、70、55 時各會回傳什麼文字？

````python
# TODO：補上三種分數範圍的回傳文字
def describe_score(score):
    if score >= 80:
        return None
    elif score >= 60:
        return None
    else:
        return None

assert describe_score(82) == "表現良好", "檢查 80 分以上的分支。"
assert describe_score(70) == "達到基本標準", "檢查 60–79 分的分支。"
assert describe_score(55) == "需要加強", "檢查 60 分以下的分支。"
print(describe_score(82), describe_score(70), describe_score(55))
````

````python
# W2 整合練習：從頭重建資料。
# 這格可在重新啟動核心後單獨執行。
from pathlib import Path
import numpy as np
import pandas as pd

random_seed = 42
n_full = 60
rng_full = np.random.default_rng(random_seed)
departments_full = ["資訊管理", "企業管理", "金融管理"]
student_ids_full = np.arange(1, n_full + 1)
chosen_departments_full = rng_full.choice(departments_full, size=n_full)
sleep_values_full = rng_full.normal(6.7, 1.1, size=n_full).round(1)
raw_screen_values_full = rng_full.normal(5.0, 1.5, size=n_full)
screen_values_full = np.clip(raw_screen_values_full, 1, 10).round(1)
raw_study_values_full = rng_full.normal(2.5, 1.0, size=n_full)
study_values_full = np.clip(raw_study_values_full, 0.5, 6).round(1)
attendance_values_full = rng_full.normal(88, 8, size=n_full)
attendance_values_full = attendance_values_full.clip(60, 100).round().astype(int)

df_full = pd.DataFrame({
    "student_id": student_ids_full,
    "department": chosen_departments_full,
    "sleep_hours": sleep_values_full,
    "screen_time_hours": screen_values_full,
    "study_hours": study_values_full,
    "attendance_rate": attendance_values_full,
})
raw_scores_full = (
    50
    + 3 * df_full["study_hours"]
    + 2 * df_full["sleep_hours"]
    - 1.2 * df_full["screen_time_hours"]
    + 0.15 * df_full["attendance_rate"]
    + rng_full.normal(0, 5, size=n_full)
)
df_full["quiz_score"] = raw_scores_full.clip(0, 100).round().astype(int)
df_full.loc[[5, 17], "sleep_hours"] = np.nan
df_full.loc[8, "screen_time_hours"] = 15.0

display(df_full.head())
print("資料大小：", df_full.shape)
display(df_full.dtypes)
display(df_full.isna().sum())
display(df_full.describe())
df_full.to_csv("student_learning_data_rebuilt.csv", index=False, encoding="utf-8-sig")

assert df_full.shape == (60, 7)
assert df_full["sleep_hours"].isna().sum() == 2
assert (df_full["screen_time_hours"] > 10).sum() == 1
assert Path("student_learning_data_rebuilt.csv").is_file()
print("整合任務檢查通過。")
````

### C. 從檔案讀取金鑰，呼叫模型

下一格用 `read_text(...).strip()` 讀取 `./api.key`，其中 `strip()` 會移除行尾換行與空白。請勿在程式中加入 `print(api_key)`。執行會送出一個 API 請求，可能產生 API 用量。

````python
from pathlib import Path
import json
from urllib.request import Request, urlopen

API_KEY_PATH = Path.cwd() / "api.key"
LLM_BASE_URL = "https://www.iai.nkust.edu.tw/aihub/v1/"
LLM_API_URL = LLM_BASE_URL.rstrip("/") + "/chat/completions"
LLM_MODEL = "Nkust"

# 這三個函式保留為可替換範本，方便日後改用其他服務。
def build_payload(prompt, model):
    return {"model": model, "messages": [{"role": "user", "content": prompt}]}

def build_headers(key):
    return {"Authorization": "Bearer " + key, "Content-Type": "application/json"}

def extract_text(response_data):
    return response_data["choices"][0]["message"]["content"]

if not API_KEY_PATH.is_file():
    print("找不到 ./api.key；請先完成上一格的金鑰檔建立步驟。")
else:
    api_key = API_KEY_PATH.read_text(encoding="utf-8").strip()
    if not api_key or "\n" in api_key:
        raise ValueError("api.key 必須只包含一行非空白金鑰。")

    teaching_prompt = "請用三句繁體中文解釋 Python 的 for 迴圈。請先給我一個提示，不要直接提供本作業的程式答案。"
    assert teaching_prompt.strip(), "請先填寫 teaching_prompt。"
    request = Request(
        LLM_API_URL,
        data=json.dumps(build_payload(teaching_prompt, LLM_MODEL)).encode("utf-8"),
        headers=build_headers(api_key),
        method="POST",
    )
    with urlopen(request, timeout=30) as response:
        response_data = json.load(response)
    print(extract_text(response_data))
    del api_key  # 使用完後移除這個變數；金鑰檔仍保留在本機
````
